```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier
```

```python
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0,1,1,0])
```

```python
model = MLPClassifier(
      hidden_layer_sizes=(2,),    # 2 neurons in hidden layer
      activation='logistic',      # Sigmoid activation
      solver='sgd',               #Gradiet Descent
      learning_rate_init=0.5,
      max_iter=5000,
      random_state=1
)
```

```python
model.fit(X,y)
prediction = model.predict(X)
```

```python
print("Actual Output")
print(y)
print("\nPredicted Output")
print(prediction)
print("\nAccuracy:", model.score(X, y) * 100,"%")
print("\nFinal Loss :", round(model.loss_,6))
```

```python
plt.figure(figsize = (7,5))
plt.plot(model.loss_curve_)
plt.title("Epoch vs Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
