```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
X = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50],
    [40, 50, 60],
    [50, 60, 70]
])

y = np.array([40, 50, 60, 70, 80])

X = X / 100
y = y / 100

X = X.reshape(5, 3, 1)

model = Sequential()
model.add(SimpleRNN(20, activation="tanh", input_shape=(3, 1)))
model.add(Dense(1))

model.compile(optimizer="adam", loss="mse")

model.fit(X, y, epochs=1000, verbose=0)

test = np.array([[60, 70, 80]]) / 100
test = test.reshape(1, 3, 1)

prediction = model.predict(test, verbose=0)

print("Predicted next value:", prediction[0][0] * 100)

print("Name: Omkar Vaidya")
print("Roll No.: 40")
```
