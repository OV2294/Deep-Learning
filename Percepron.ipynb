```python
import numpy as np
```

```python
class Perceptron:
    def __init__(self, input_size, learning_rate=1):
        self.weights = np.zeros(input_size)
        self.bias = 0
        self.learning_rate = learning_rate
    def step_function(self, x):
        if x > 0:
            return 1
        elif x == 0:
            return 0
        else:
            return -1
    def predict(self, inputs):
        total = np.dot(inputs, self.weights) + self.bias
        return self.step_function(total)
    def train(self, inputs, targets, max_epochs=100):
        for epoch in range(max_epochs):
            errors = 0
            for i in range(len(inputs)):
                prediction = self.predict(inputs[i])
                error = targets[i] - prediction
                if error !=0:
                    self.weights += self.learning_rate * targets[i] * inputs[i]
                    self.bias += self.learning_rate * targets[i]
                    errors += 1
            print(f"Epoch {epoch+1}: errors = {errors}")
            if errors == 0:
                print(f"Training stopped after {epoch+1} epochs.")
                break
```

```python
inputs = np.array([[1,1],[1,-1],[-1,1],[-1,-1]])
targets = np.array([1,-1,-1,-1])

perceptron = Perceptron(input_size=2)
perceptron.train(inputs, targets)

print("AND Gate Predictions: ")
for input_data in inputs:
    print(f"Input: {input_data}, Prediction: {perceptron.predict(input_data)}")
```
